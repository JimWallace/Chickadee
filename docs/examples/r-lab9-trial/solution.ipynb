Predicting hypertension (`high_bp`) from NHANES adult data. This is the instructor answer key used to validate the test suite.

```r
source("lab9_helpers.R")   # load_data(), build_cohort(), evaluate_model(), FEATURES
```

```r
cohort <- build_cohort(20240617)
cat("Cohort size:", nrow(cohort), "\n")
```

```r
summarize <- function(cohort) {
    # Cohort size, mean age, and hypertension rate.
    list(
        n = nrow(cohort),
        mean_age = mean(cohort$age),
        hypertension_rate = mean(cohort$high_bp)
    )
}

summarize(cohort)
```

```r
strongest_predictor <- function(cohort) {
    # Name of the feature most strongly correlated (in absolute value) with high_bp.
    cors <- vapply(FEATURES, function(f) cor(cohort[[f]], cohort$high_bp), numeric(1))
    names(which.max(abs(cors)))
}

strongest_predictor(cohort)
```

```r
select_features <- function() {
    # Features that actually predict high blood pressure.
    c("age", "waist", "bmi")
}

my_features <- select_features()
cat("Accuracy:", evaluate_model(my_features, load_data()), "\n")
```
